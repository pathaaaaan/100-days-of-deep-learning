## Level 4 — apply()

```python
df["Average_Score"] = df[
    ["Math_Score", "Programming_Score"]
].mean(axis=1)
```

```python
df["Average_Score"]
```

```python
def performance(score):

    if score >= 90:
        return "Excellent"

    elif score >= 75:
        return "Good"

    elif score >= 60:
        return "Average"

    else:
        return "Poor"
```

```python
df["Performance"] = df["Average_Score"].apply(
    performance
)
```

## Level 8 — merge()

```python
scholarship_data = [
    [101, 50000],
    [103, 75000],
    [105, 60000],
    [107, 80000],
    [109, 45000],
    [111, 90000],
    [113, 70000]
]

scholarship_df = pd.DataFrame(
    scholarship_data,
    columns=["ID", "Scholarship"]
)

scholarship_df
```

```python
result = pd.merge(
    df,
    scholarship_df,
    on="ID",
    how="left"
)
"""
Why left?
Because we want to keep every student, even if they don't have a scholarship.
"""
result
```

## Level 9 — Different Types of Merge


```python
# inner : Only students present in both DataFrames
pd.merge(
    df,
    scholarship_df,
    on="ID",
    how="inner"
)
```

```python
# left : Everything from df.
pd.merge(
    df,
    scholarship_df,
    on="ID",
    how="left"
)
```

```python
# Right : everything from scholarship_df
pd.merge(
    df,
    scholarship_df,
    on="ID",
    how="right"
)
```

```python
# outer : Everything from both.
pd.merge(
    df,
    scholarship_df,
    on="ID",
    how="outer"
)
```
