## Handling Missing Values

```python
import pandas as pd
import numpy as np

data = [
    [1, "Hasnain", 21, "Dehradun", 45000, "Male"],
    [2, "Rahul", 22, "Delhi", 55000, "Male"],
    [3, "Priya", np.nan, "Mumbai", 40000, "Female"],
    [4, "Aman", 23, None, 60000, "Male"],
    [5, "Sneha", 21, "Bangalore", np.nan, "Female"],
    [6, None, 24, "Delhi", 52000, "Male"],
    [7, "Riya", np.nan, "Mumbai", 48000, "Female"],
    [8, "Arjun", 25, "Pune", 70000, None],
    [9, "Vikash", 22, "Delhi", 55000, "Male"],
    [10, "Neha", 20, "Mumbai", 35000, "Female"],
    [11, "Karan", 26, "Bangalore", 80000, "Male"],
    [12, "Anjali", 23, None, 62000, "Female"],
    [13, "Rohit", 27, "Delhi", np.nan, "Male"],
    [14, "Pooja", 22, "Mumbai", 45000, None],
    [15, "Sameer", 24, "Pune", 58000, "Male"],
    [16, "Meena", np.nan, "Delhi", 50000, "Female"],
    [17, "Arjun", 25, "Pune", 70000, "Male"],  # duplicate
    [18, "Raj", 21, "Delhi", -5000, "Male"],     # invalid salary
    [19, "Simran", 29, "Mumbai", 90000, "Female"],
    [20, "Aman", 23, "Delhi", 60000, "Male"],
]

columns = ["ID", "Name", "Age", "City", "Salary", "Gender"]

df = pd.DataFrame(data, columns=columns)

df
```

## Missing Values

```python
df.isnull().sum()
```

```python
df.isnull()
```

```python
# rows containing missing values
df.isnull().any(axis = 1)
```

```python
# drop rows witg null values
df.dropna()
```

```python
# fill missing values
df["Age"].fillna(df["Age"].mean())
```

```python
# fill categorilcal nulls
df["City"].fillna("Unknown")
```

```python
# find duplicates
df.duplicated()
# df[df.duplicated()]
```

```python
df[df.duplicated()]
```

```python
# invalid values
df[df["Salary"] < 0]
```

```python
# replace invalid salaries
df.loc[df["Salary"] < 0, "Salary"] = np.nan
```
