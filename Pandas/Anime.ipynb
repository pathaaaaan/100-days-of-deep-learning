```python
def extract_episodes(text):
    check = False;
    data = ""
    for i in text:
        if i == ')':
            break;
        if i == '(':
            check = True
            continue
        if check:
            data += i
    return data;
```

```python
df["Epidoes"] = df["Title"].apply(extract_episodes)
```

```python
df['Epidoes'] = df['Epidoes'].str.replace(' eps', "")
```

```python
def extract_timestamp(txt):
    check = False
    ans = ""
    l = 0
    for i in txt:
        if i == ')':
            check = True;
            continue
        if check:
            ans += i;
            l += 1;
        if l == 19:
            return ans;
    return ans;
```

```python
df['TimeLine'] = df["Title"].apply(extract_timestamp)
```
