```python
import pandas as pd
import numpy as np

# ============================================================
# DATASET
# ============================================================

data = [
    [1, "Hasnain", 21, "Delhi", 45000, "Male", "Yes"],
    [2, "Rahul", 22, "Mumbai", 55000, "Male", "No"],
    [3, "Priya", np.nan, "Delhi", 40000, "Female", "Yes"],
    [4, "Aman", 23, None, 60000, "Male", "Yes"],
    [5, "Sneha", 21, "Bangalore", np.nan, "Female", "No"],
    [6, None, 24, "Delhi", 52000, "Male", "Yes"],
    [7, "Riya", np.nan, "Mumbai", 48000, "Female", "Yes"],
    [8, "Arjun", 25, "Pune", 70000, None, "No"],
    [9, "Vikash", 22, "Delhi", 55000, "Male", "Yes"],
    [10, "Neha", 20, "Mumbai", 35000, "Female", "No"],
    [11, "Karan", 26, "Bangalore", 80000, "Male", "Yes"],
    [12, "Anjali", 23, None, 62000, "Female", "Yes"],
    [13, "Rohit", 27, "Delhi", np.nan, "Male", "No"],
    [14, "Pooja", 22, "Mumbai", 45000, None, "Yes"],
    [15, "Sameer", 24, "Pune", 58000, "Male", "No"],
    [16, "Meena", np.nan, "Delhi", 50000, "Female", "Yes"],
    [17, "Arjun", 25, "Pune", 70000, "Male", "No"],
    [17, "Arjun", 25, "Pune", 70000, "Male", "No"],
    [18, "Raj", 21, "Delhi", -5000, "Male", "Yes"],
    [19, "Simran", 29, "Mumbai", 90000, "Female", "Yes"],
    [20, "Aman", 23, "Delhi", 60000, "Male", "Yes"],
    [21, "Kabir", 35, "Delhi", 120000, "Male", "No"],
    [22, "Nisha", 19, "Mumbai", 25000, "Female", "Yes"],
    [23, "Ravi", 22, "Delhi", 55000, "M", "Yes"],
    [24, "Pinky", 24, "Mumbai", 60000, "F", "No"],
    [25, "Mohit", -2, "Pune", 45000, "Male", "Yes"],
    [26, "Sara", 21, "Delhi", 48000, "Female", None],
    [27, "Dev", 28, "Bangalore", 75000, "Male", "Yes"],
    [28, "Isha", 22, "Delhi", 50000, "Female", "No"],
    [29, "Rohan", 23, "Mumbai", 55000, "Male", "Yes"],
    [30, "Kriti", 21, "Bangalore", 65000, "Female", "Yes"],
]

columns = ["ID", "Name", "Age", "City", "Salary", "Gender", "Remote"]

df = pd.DataFrame(data, columns=columns)
df
```

### Level 1 — Inspection

```python
# Q1. Number of rows and columns
print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
print('-'*70)

# Q2. First 5 rows
print(df.head())
print('-'*70)


# Q3. Last 5 rows
print(df.tail())
print('-'*70)


# Q4. Data types
print(df.dtypes)
print('-'*70)


# Q5. Statistical summary
print(df.describe())
print('-'*70)


# Q6. Number of unique cities
print(f"Number of Unique cities: {df['City'].nunique()}")
print('-'*70)


# Q7. Unique cities
print(df["City"].unique())
print('-'*70)


# Q8. Number of employees in each city
print("Numbr of employees in each city\n",df["City"].value_counts())
print('-'*70)
```

## Level 2 : Missing Values

```python
# Q9. Missing values in every column
print(df.isnull().sum())


# Q10. Rows containing at least one missing value
print(df[df.isnull().any(axis=1)])


# Q11. Number of missing Age values
print(f"Missing Age: {df['Age'].isnull().sum()}")


# Q12. Fill missing Age with mean
mean_age = df["Age"].mean()

df["Age"] = df["Age"].fillna(mean_age)

print(f"Age filled with mean: {mean_age:.2f}")


# Q13. Fill missing City
df["City"] = df["City"].fillna("Unknown")

print("Missing City values replaced with 'Unknown'.")


# Q14. Fill missing Gender
df["Gender"] = df["Gender"].fillna("Unknown")

print("Missing Gender values replaced with 'Unknown'.")


# Q15. Remove rows where Name is missing
df = df.dropna(subset=["Name"])

print(f"Rows after removing missing names: {len(df)}")


# Q16. Remove rows where all values are missing
df = df.dropna(how="all")

print(f"Rows remaining: {len(df)}")
```

### Level 3 — Filtering

```python
# Q17. Age greater than 25
result = df[df["Age"] > 25]

print("Age > 25:")
print(result)


# Q18. Employees from Delhi
result = df[df["City"] == "Delhi"]

print("Employees from Delhi:")
print(result)


# Q19. Delhi AND salary > 50000
result = df[
    (df["City"] == "Delhi") &
    (df["Salary"] > 50000)
]

print("Delhi employees earning > ₹50,000:")
print(result)


# Q20. Delhi OR Mumbai
result = df[
    (df["City"] == "Delhi") |
    (df["City"] == "Mumbai")
]

print("Employees from Delhi or Mumbai:")
print(result)


# Q21. Salary between 40000 and 70000
result = df[
    df["Salary"].between(40000, 70000)
]

print("Salary between ₹40,000 and ₹70,000:")
print(result)


# Q22. Age between 20 and 25
result = df[
    df["Age"].between(20, 25)
]

print("Age between 20 and 25:")
print(result)


# Q23. Employees NOT from Delhi
result = df[df["City"] != "Delhi"]

print("Employees not from Delhi:")
print(result)


# Q24. Female employees earning > 50000
result = df[
    (df["Gender"] == "Female") &
    (df["Salary"] > 50000)
]

print("Female employees earning > ₹50,000:")
print(result)
```

## Level 4 — Duplicates

```python
# Q25. Find duplicate rows
duplicates = df[df.duplicated()]

print("Duplicate rows:")
print(duplicates)


# Q26. Count duplicate rows
count = df.duplicated().sum()

print(f"Number of duplicate rows: {count}")


# Q27. Remove duplicate rows
df = df.drop_duplicates()

print(f"Rows after removing duplicates: {len(df)}")


# Q28. Find duplicate IDs
duplicates = df[df["ID"].duplicated()]

print("Duplicate IDs:")
print(duplicates)


# Q29. Remove duplicate IDs
df = df.drop_duplicates(
    subset=["ID"],
    keep="first"
)

print("Duplicate IDs removed.")
```

## Level 5 — Invalid Data

```python
# Q30. Find negative salaries
result = df[df["Salary"] < 0]

print("Negative salaries:")
print(result)


# Q31. Find negative ages
result = df[df["Age"] < 0]

print("Negative ages:")
print(result)


# Q32. Replace negative salary with NaN
df.loc[
    df["Salary"] < 0,
    "Salary"
] = np.nan

print("Negative salaries replaced with NaN.")


# Q33. Replace negative age with NaN
df.loc[
    df["Age"] < 0,
    "Age"
] = np.nan

print("Negative ages replaced with NaN.")


# Q34. Fill missing salary with median
median_salary = df["Salary"].median()

df["Salary"] = df["Salary"].fillna(
    median_salary
)

print(f"Missing salaries filled with median: ₹{median_salary:,.2f}")
```

## Level 6 — Data Transformation

```python
# Q35. Create Annual Salary
df["Annual_Salary"] = df["Salary"] * 12

print(df[["Name", "Salary", "Annual_Salary"]])


# Q36. Create Salary Level
def salary_level(salary):
    if salary < 50000:
        return "Low"
    elif salary <= 80000:
        return "Medium"
    else:
        return "High"


df["Salary_Level"] = df["Salary"].apply(
    salary_level
)

print(df[["Name", "Salary", "Salary_Level"]])


# Q37. Create Age Group
def age_group(age):
    if age < 22:
        return "Young"
    elif age <= 25:
        return "Adult"
    else:
        return "Senior"


df["Age_Group"] = df["Age"].apply(
    age_group
)

print(df[["Name", "Age", "Age_Group"]])


# Q38. Standardize Gender
df["Gender"] = df["Gender"].replace({
    "M": "Male",
    "F": "Female"
})

print(df["Gender"].value_counts())
```

## Level 7 — Sorting

```python
# Q39. Salary: Low → High
result = df.sort_values(
    by="Salary",
    ascending=True
)

print(result[["Name", "Salary"]])


# Q40. Salary: High → Low
result = df.sort_values(
    by="Salary",
    ascending=False
)

print(result[["Name", "Salary"]])


# Q41. City ascending + Salary descending
result = df.sort_values(
    by=["City", "Salary"],
    ascending=[True, False]
)

print(result[["Name", "City", "Salary"]])
```

## Level 8 — loc and iloc

```python
# Q42. Delhi employees: Name, Age, Salary
result = df.loc[
    df["City"] == "Delhi",
    ["Name", "Age", "Salary"]
]

print(result)


# Q43. First 5 rows using iloc
result = df.iloc[0:5]

print(result)


# Q44. Rows 5–10 and columns 1–4
result = df.iloc[5:11, 1:5]

print(result)


# Q45. Age > 22 AND Salary > 50000
result = df.loc[
    (df["Age"] > 22) &
    (df["Salary"] > 50000)
]

print(result)
```

## Level 9 — GroupBy

```python
# Q46. Average salary by city
result = df.groupby("City")["Salary"].mean()

print("Average Salary by City:")

for city, salary in result.items():
    print(f"{city:<12} ₹{salary:,.2f}")


# Q47. Maximum salary by city
result = df.groupby("City")["Salary"].max()

print("\nMaximum Salary by City:")

for city, salary in result.items():
    print(f"{city:<12} ₹{salary:,.0f}")


# Q48. Minimum salary by city
result = df.groupby("City")["Salary"].min()

print("\nMinimum Salary by City:")

for city, salary in result.items():
    print(f"{city:<12} ₹{salary:,.0f}")


# Q49. Employee count by city
result = df["City"].value_counts()

print("\nEmployees by City:")

for city, count in result.items():
    print(f"{city:<12} {count} employees")


# Q50. Average age by gender
result = df.groupby("Gender")["Age"].mean()

print("\nAverage Age by Gender:")

for gender, age in result.items():
    print(f"{gender:<12} {age:.2f} years")


# Q51. Total salary by city
result = df.groupby("City")["Salary"].sum()

print("\nTotal Salary by City:")

for city, salary in result.items():
    print(f"{city:<12} ₹{salary:,.0f}")


# Q52. City with highest average salary
result = df.groupby("City")["Salary"].mean()

city = result.idxmax()
salary = result.max()

print(
    f"\nHighest average salary: "
    f"{city} → ₹{salary:,.2f}"
)
```
