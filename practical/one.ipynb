## GRU

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense

seq = np.array([2, 3, 7, 11, 16], dtype=float)

seq = seq / 16

X = seq[:-1]
y = seq[1:]

X = X.reshape((4, 1, 1))

model = Sequential()
model.add(GRU(50, activation="tanh", input_shape=(1, 1)))
model.add(Dense(1))

model.compile(optimizer="adam", loss="mse")
model.fit(X, y, epochs=500, verbose=0)

test = np.array([[16]]) / 16
test = test.reshape((1, 1, 1))

pred = model.predict(test)
pred = pred * 16

print("Prediction", pred)
```

## ANN


```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# sequence
seq = np.array([5, 10, 15, 20], dtype=float)

# create dataset
X = seq[:-1]
y = seq[1:]

# reshape for ANN
X = X.reshape((3, 1))

# build ANN model
model = Sequential()

model.add(Dense(10, activation="relu", input_dim=1))
model.add(Dense(1))

# compile model
model.compile(optimizer="adam", loss="mse")

# train
model.fit(X, y, epochs=500, verbose=0)

# test case (predict after 20)
test = np.array([[20]])

prediction = model.predict(test)

print("Prediction:", prediction)
```
